```python
from sklearn.feature_extraction.text import CountVectorizer

#  Créer une liste de phrases
phrases = [
    "Le football est un sport populaire",
    "Le basketball est un sport collectif",
    "Le football et le basketball sont des sports"
]


vectorizer = CountVectorizer()


X = vectorizer.fit_transform(phrases)

#  Afficher les mots extraits
print("Liste des mots :")
print(vectorizer.get_feature_names_out())

#  Afficher la matrice
print("Matrice BoW :")
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Créer le vectoriseur TF-IDF
vectorizer = TfidfVectorizer()

# 2. Transformer les phrases
X = vectorizer.fit_transform(phrases)

# 3. Afficher les mots extraits
print("Liste des mots :")
print(vectorizer.get_feature_names_out())

# 4. Afficher la matrice TF-IDF
print("Matrice TF-IDF :")
print(X.toarray())
```
